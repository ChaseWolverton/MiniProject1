# Mini Project 1: Pokémon Card Value Analysis

I plan to analyze Pokémon card data to see how the value of Pokémon cards has changed over time. I am thinking about comparing different card sets, rarities, Pokémon, and release years to find patterns that might affect how much a card is worth.

For my data, I plan to use publicly available Pokémon card databases and historical card prices. Some comparisons I am thinking of are comparing older cards to newer cards, comparing the prices of different rarities, and looking at how the value of certain cards has changed over time.

I chose Pokémon cards because I am personally interested in collecting and selling them, and I think it would be interesting to see what factors have the biggest impact on a card's value and how prices have changed throughout the history of Pokémon cards.
